```python
import google.generativeai as genai

from google.colab import userdata
GOOGLE_API_KEY = userdata.get("GOOGLE_API_KEY")
genai.configure(api_key=GOOGLE_API_KEY)
model = genai.GenerativeModel("gemini-3.6-flash")
```

```python
response = model.generate_content("こんにちははやてです")
print(response.text)
```

```python
chat = model.start_chat(
    history=[
        {"role":"user","parts":["こんにちは！私ははやてと言います"]},
        {"role":"model","parts":["こんにちははやてさん初めまして！　何かお手伝いしましょうか"]}
    ]
)
response = chat.send_message("私の名前が分かりますか？")
print(response.text)
```

```python
print("ストリーム")
```

```python
#modelの生成までは上のものを流用
response= model.generate_content(
    "こんにちは、私はオカヤマと言います。",
    stream=True
)
# 応答を順次表示する
for chunk in response:
  print(chunk.text, end="", flush=True)
```

```python
#modelの生成までは上のものを流用
response= model.generate_content(
    "昔々あるところにおじいさんとおばあさんが居ました。登場人物を次のJSON形式で出力してください。{'people:['aaa','bbb']'}",
    generation_config={"response_mime_type":"application/json"}
)
print(response.text)
```

```python
#　必要なライブラリの追加　import
import requests
from PIL import  Image
from io import  BytesIO

image_url="https://raw.githubusercontent.com/yoshidashingo/langchain-book/main/assets/cover.jpg"
response_img=requests.get(image_url)
img= Image.open(BytesIO(response_img.content))

# modelの生成は上の部分を流用
response = model.generate_content([
    "この画像を説明してください",
    img
])
print(response.text)
```
